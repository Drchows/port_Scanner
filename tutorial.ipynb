```python
import socket

host = "127.0.0.1"
port = 8000

try: 
    sock = socket.create_connection((host, port), timeout=2)
    print("Connection succesful!")
    print(f"Port {port} is OPEN")

except OSerror:
    print(f"COuld not connect to port {port}")
```

```python
with sock:
    request = (
        "HEAD / HTTP/1.0\r\n"
        "Host: localhost\r\n"
        "\r\n"
    ).encode("utf-8")

    sock.sendall(request)

    web_response = []
    
    while True:
        response = sock.recv(1024)
        if not response:
            break
        web_response.append(response)
        
full_response = b"".join(web_response)
print(full_response.decode("utf-8", errors = "replace"))
```
